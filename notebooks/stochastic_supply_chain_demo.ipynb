# Stochastic supply-chain simulation

This notebook turns a small discrete-time reaction model into a Monte Carlo experiment. The point is not to predict a real company; it is to make randomness, state transitions, and uncertainty visible.

```python
from pathlib import Path
import sys
import pandas as pd
import matplotlib.pyplot as plt

sys.path.insert(0, str(Path('..').resolve()))
from stochastic_supply_chain import run_experiment

results = run_experiment(runs=500, initial_a=100, initial_b=100, steps=1_000, completion_probability=0.9)
summary = pd.DataFrame([vars(state) for state in results])
summary.describe().round(2)
```

## What randomness changes

The expected process is simple, but the final number of completed products varies from run to run. Repeating the same experiment gives an empirical distribution rather than a single misleading answer.

```python
ax = summary['product_c'].plot(kind='hist', bins=18, edgecolor='white', title='Distribution of completed products')
ax.set_xlabel('product C units at the end of the run')
plt.show()
```

## Sensitivity to completion probability

A small parameter sweep shows how process reliability changes expected output. This is a useful pattern for simulation studies: vary one assumption, hold the rest constant, and compare distributions.

```python
rows = []
for probability in [0.5, 0.7, 0.9, 1.0]:
    trial = run_experiment(runs=300, initial_a=100, initial_b=100, steps=1_000, completion_probability=probability)
    rows.append({'completion_probability': probability, 'mean_product_c': sum(s.product_c for s in trial) / len(trial)})
pd.DataFrame(rows)
```

## Takeaway

The model is small enough to inspect line by line, yet it demonstrates core stochastic-process habits: define the state, make transitions explicit, repeat with controlled seeds, and report a distribution instead of hiding uncertainty behind one run.
